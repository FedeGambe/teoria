## Abstract Base Class (ABC)

In Python, una Abstract Base Class (ABC) è una classe che non può (o non dovrebbe) essere istanziata direttamente e che serve come modello per altre classi.

Definisce uno “scheletro” di metodi che le classi figlie devono obbligatoriamente implementare.

**A cosa serve?**

- Definire un’**interfaccia comune**

- Forzare una struttura coerente tra classi correlate

- Rendere il codice più chiaro e robusto (principio di progettazione OOP)

**Come si crea in Python?**

Si usa il modulo standard `abc` (Abstract Base Classes).

```python
from abc import ABC, abstractmethod

class Animale(ABC):     # Animale è una classe astratta:

    @abstractmethod     # è un decoratore, serve a marcare un metodo di una classe astratta come “astratto”;
    def verso(self):    # il metodo verso() è astratto.
        print("Sto facendo qualcosa di base...")  # implementazione di default
        pass
                        # Qualsiasi classe che eredita da Animale deve implementare verso()
```

```python
class Cane(Animale):

    def verso(self):
        return "Bau!"

c = Cane()
print(c.verso())  # Bau!
```

Una ABC può avere implementazioni di metodi anche se astratti: il metodo astratto può contenere codice che le sottoclassi possono richiamare con `super()`.

`super()` è una funzione che permette a una sottoclasse di chiamare metodi della classe genitore (la classe da cui eredita)


```python
class Cane(Animale):
    def verso(self):
        super().verso()  # chiama la versione della classe astratta
        return "Bau!"

c = Cane()
print(c.verso())
```

```python
# Se si prova invece a richiamare la classe ABC:
# Si ottterà un errore, perché non puoi istanziare una classe astratta.
a = Animale()
```

**Concetto chiave**

Una Abstract Base Class:

- Può avere metodi normali

- Può avere metodi astratti

- Serve a definire un contratto che le sottoclassi devono rispettare

| Classe normale          | Abstract Base Class                |
| ----------------------- | ---------------------------------- |
| Può essere istanziata   | Non può essere istanziata          |
| I metodi sono opzionali | I metodi astratti sono obbligatori |


### 2. Tipi di metodi astratti

Oltre ai metodi normali e astratti (`@abstractmethod`), Python permette anche:

- metodi astratti di classe `(@classmethod`)

- metodi astratti statici (`@staticmethod`)

- proprietà astratte (`@property`)

```python
from abc import ABC, abstractmethod

class Animale(ABC):

    @abstractmethod
    def verso(self):  # metodo di istanza astratto
        pass

    @classmethod
    @abstractmethod
    def info_classe(cls):  # metodo di classe astratto
        pass

    @staticmethod
    @abstractmethod
    def numero_zampe():  # metodo statico astratto
        pass

    @property
    @abstractmethod
    def colore(self):  # proprietà astratta
        pass
```

Se si prova a creare una sottoclasse di `Animale` bisogna implementare tutti e quattro, altrimenti Python  darà errore quando si prova ad istanziarla.
