## Dataclass

Le dataclass servono a creare classi **“contenitori di dati**” senza scrivere troppo codice boilerplate (cioè `__init__`, `__repr__`, `__eq__`, ecc.).

Bisogna importare dataclass da dataclasses, e mettere un decoratore `@dataclass` prima della classe.

```python
from dataclasses import dataclass

@dataclass
class Punto:
    x: int
    y: int
```

Quindi, in questo caso non serve un costruttore, perché è automatico, ha già una rapprensetazione leggibile (`__repr__`) e l'equiquivalenza tra gli oggetti (`__eq__`).

#### Valori di default

Si possono passare dei valori di default, dando agli attributi dei valori iniziali,così se non lo si passa al momento di creare l’oggetto, Python usa quello di default:

```python
@dataclass
class Punto:
    x: int = 0
    y: int = 0
    z: int = 0

print(Punto())
print(Punto(5,2))
```

#### `field()`

Viene usato per comportamenti avanzati, permette di dire come Python deve trattare un attributo.

💡 Utile per dati sensibili (come password).

```python
from dataclasses import dataclass, field

@dataclass
class Persona:
    nome: str
    password: str = field(repr=False, compare=False)

    # repr=False → non apparirà quando fai print().
    # compare=False → non sarà usato per confrontare oggetti con == .
```

```python
p = Persona("Luca", "1234")
print(p)  # Persona(nome='Luca') → password non si vede
```

#### Ordinabilità

Se si vuole confrontare o ordinare degli oggetti, bisdogna inserire nel decoratore `(order=True)`.

Per esempio ordinare dei punti per le cordinate:

```python
@dataclass(order=True)
class Punto:
    x: int
    y: int
```

- Python crea automaticamente i metodi: `<`, `<=`, `>`, `>=`.

- Ordina prima per `x`, poi per `y` se `x` è uguale.

```python
p1 = Punto(1, 2)
p2 = Punto(2, 1)
print(p1 < p2)  # True
```

#### Immutabilità

Viene usate se si vuole che gli oggetti n**on cambino dopo la loro creazione**.

Per utilizzare questa caratterisica bisogna impostare nel decoratore `(frozen=True)`.

```python
@dataclass(frozen=True)
class Punto:
    x: int
    y: int
```

Quando `frozen=True` viene usato il metodo spciale `__hash__` in automatico, serve a rendere la classe hashable, cioè utilizzabile come chiave di un dizionario o in un set.

```python
p = Punto(1, 2)
p
```

Se si vuole modificare `x`, per esempio, si ottiene un errore: `FrozenInstanceError`.
Nel quale dice che non si possono assegnare valori a `x`.

```python
p.x = 5
p
```

#### Metodi personalizzati

Anche se Python crea automaticamente `__init__` e `__repr__`, si possono comuqnue aggiungere funzioni

```python
@dataclass
class Punto:
    x: int
    y: int

    def distanza(self):
        return (self.x**2 + self.y**2)**0.5

p = Punto(3, 4)
print(f"La diztanza tra: {p}, è di: {p.distanza()}")  # 5.0
```

#### Validazione post inizializzazione

Viene usato `__post_init__` che non è automatico come `__init__` ma **opzionale**. È un metodo speciale che viene chiamato subito dopo `__init__`.

Utile per validazioni o calcoli extra sugli attributi.

```python
@dataclass
class Punto:
    x: int
    y: int

    # In questo caso si vuole mettere una validazione che controlli se le cordinate immesse sono negativa,
    # Se si, manda un messaggio all'utente
    def __post_init__(self):
        if self.x < 0 or self.y < 0:
            print("Attenzione: coordinate negative")
```

```python
p1 = Punto(1,3)
p1
```

```python
p2 = Punto(-1,3)
p2
```
