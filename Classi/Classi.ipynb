## Classi

Le classi sono il cuore della Programmazione Orientata agli Oggetti (OOP) in Python.
Permettono di raggruppare dati (attributi) e le funzioni che operano su quei dati (metodi) in un'unica unità logica.

Si usa `class` per definire un modello (o blueprint) per creare oggetti (istanze).
- **Classe**: Definisce la struttura (gli attributi) e il comportamento (i metodi) che gli oggetti di quel tipo avranno.
- **Oggetto** (Istanza): È una realizzazione concreta della classe, un'entità individuale creata
seguendo quel modello

All’interno della classe si possono definire due cose principali:
- **Attributi** → i dati che descrivono l’oggetto (come colore, marca, età).
- **Metodi** → funzioni che fanno agire l’oggetto (come guidare, fermarsi, cambiare colore).

*Si usa la parola chiave class seguita dal nome della classe (per convenzione, in PascalCase o
CamelCase).*

### 1. **I metodi Speciali**

##### 1.1. **Costruttore** `__init__`

Il metodo speciale `__init__`:
- __init__ è un metodo speciale che Python chiama costruttore.
- Viene eseguito automaticamente quando si
crea un nuovo oggetto della classe
- Serve per inizializzare l’oggetto, cioè dare i valori iniziali agli **attributi di istanza**.

````python
def __init__(self, parametro1, parametro2, ...):
    self.attributo1 = parametro1
    self.attributo2 = parametro2
````

- `self` rappresenta l’oggetto stesso. Ogni volta che si crea un oggetto, `self` permette ai metodi di accedere e modificare i suoi **attributi di istanza**.

  `self` è l’istanza corrente della classe su cui stai lavorando in quel momento.  

````python
class Cane:                     # Classe / Modello
    def __init__(self, nome, eta):
        self.nome = nome        # Attributo d'istanza = nome
        self.eta = eta          # Attributo d'istanza = età
                                # gli attributi dentro __init__ vengono salvati dentro l'oggetto specifico

    def abbaia(self):           # Metodo
        print(f"{self.nome} dice: Bau!")


mio_cane = Cane("Fido", 3)      # Istanza / Oggetto
mio_cane.abbaia()
````

Quando si scrive:

```
mio_cane = Cane("Fido", 3)
```
si crea un nuovo oggetto vuoto, poi viene chiamato `__init__`, in fine passa automatichamente l'oggetto appena creato come primo paramentro (quel primo parametro è proprio self).

Dentro `__init__`, `self` è proprop `mio_cane`, non è qualcosa di speciale: è semplicemente il riferimento all’oggetto appena creato.



#### 1.2. Altri metodi speciali

Oltre a `__init__`, ci sono altri metodi “magici” che fanno comportare gli oggetti in modo speciale:
- `__str__`: 	Definisce come l’oggetto appare se lo stampi.
- `__repr__`:	Rappresentazione ufficiale per debug.
- `__len__`:	Permette di usare `len(obj)`.
- `__eq__`:	Permette di confrontare oggetti con `==`.
- `__add__`	: Sommare oggetti con `+`.
- `__mul__` : Moltiplicare oggetti con `*`.
- `__getitem__` : Permette `obj[index]` come lista (Accedere a un elemento).
- `__setitem__` : Permette di assegnare valori con `obj[index] = x`, modifica elementi interni.

`__str__`

- Serve a definire come appare l’oggetto quando lo stampi con `print()`.
- Senza `__str__`, Python stampa qualcosa del tipo `<__main__.Cane object at 0x...>`.

````python
class Cane:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def __str__(self):
        return f"Cane {self.nome}, {self.eta} anni"

fido = Cane("Fido", 3)
print(fido)  # Cane Fido, 3 anni
````

`__repr__`

- Serve a dare una rappresentazione “ufficiale” dell’oggetto, utile per debug.

- Differenza con `__str__`: `__str__` è più leggibile per l’utente, `__repr__` è più dettagliato per sviluppatori.

````python
class Cane:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def __repr__(self):
        return f"Cane('{self.nome}', {self.eta})"

fido = Cane("Fido", 3)
fido  # Cane('Fido', 3) in console
````

`__len__`

- mPermette di usare len(obj) su un oggetto.

- Utile se la classe contiene una collezione interna (lista, dizionario, ecc.).

````python
class Squadra:
    def __init__(self, giocatori):
        self.giocatori = giocatori

    def __len__(self):
        return len(self.giocatori)

s = Squadra(["Luca", "Anna", "Marco"])
print(len(s))  # 3
````

`__eq__`

- Permette di usare `==` per confrontare oggetti.

- Senza `__eq__`, `==` confronta solo l’indirizzo in memoria, non il contenuto.

````python
class Cane:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def __eq__(self, altro):
        return self.nome == altro.nome and self.eta == altro.eta

fido1 = Cane("Fido", 3)
fido2 = Cane("Fido", 3)
fido3 = Cane("Buddy", 2)

print(fido1 == fido2)  # True
print(fido1 == fido3)  # False
````

### 2. **Attributi di classe**

Ci sono due tipologie di attributi che si possono trovare in una classe:
- **Attributo di istanza**: appartiene a un singolo oggetto (`self.nome`)

- **Attributo di classe**: appartiene a tutta la classe, condiviso tra tutti gli oggetti

````python
class Persona:
    specie = "umano"        # Attributo di classe

    def __init__(self, nome, eta):
        self.nome = nome    # Attributo di istanza
        self.eta = eta

# - Ogni Persona avrà il suo nome e eta
# - Tutte le Persona condividono specie
````

Ogni oggetto ha accesso all'attributo `specie` e  può leggerlo:

````python
p1 = Persona("Luca", 30)
p2 = Persona("Anna", 25)

print(p1.specie)
print(p2.specie)
````

Si può creare un attributo di istanza solo per un oggetto senza cambaire l'attributo di classe:

````python
p1.specie = "extraterrestre"
print(p1.specie)
print(p2.specie)
````

### 3. **I metodi di istanza**

Una volta passati gli attributi al costruttore `__init__`, si possono aggiungere gli altri metodi, ovvero le funzioni legate all'oggetto.

I tutti i metodi nella classe, bisogna passare `self`, perché il metodo deve sapere a quale oggetto appartiene. Così può accedere agli attributi del costruttore con `self.attributo1`.

Dentro la classe si possono avere più metodi, e tutti avranno come primo parametro `self` se devono lavorare sui dati dell’oggetto.

````python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def saluta(self): # <- self per richimare l'appartenenza al costruttore
        print(f"Ciao, sono {self.nome} e ho {self.eta} anni")

p = Persona("Luca", 30)
p.saluta()
````

1) `class NomeClasse:` → definisce il modello

2) `__init__(self, ...)` → inizializza l’oggetto

3) `self.attributo = valore` → salva dati nell’oggetto (attributo)

4) Metodi con `self` → agiscono sui dati dell’oggetto

5) `p = NomeClasse(...)` → crea l’oggetto

### 4. **I metodi di classe e statici**

Al contrario dei metodi di istanza  che operano su un'istanza specifica, i metodi di classe e statici operano diversamente.

| Tipo              | Primo parametro | Accesso agli attributi | Decoratore      | Uso principale                                             |
| ----------------- | --------------- | ---------------------- | --------------- | ---------------------------------------------------------- |
| Metodo di istanza | `self`          | Istanza (`self`)       | Nessuno         | Operazioni su singola istanza                              |
| Metodo di classe  | `cls`           | Classe (`cls`)         | `@classmethod`  | Operazioni sull’intera classe, factory methods             |
| Metodo statico    | Nessuno         | Nessuno                | `@staticmethod` | Funzioni indipendenti ma collegate logicamente alla classe |


#### 4.1. Metodi di classe

Sono metodi che operano sulla classe e non sull'istanza.
- Usano un decoratore `@classmethod`.
- Il primo paramentro non è `self`, ma `cls` (che rappresenta la classe stessa).
- Possono leggere o modificare attributi di classe condivisi da tutti gli oggetti, ma non possono accedere agli attributi dell’istanza (`self` non esiste qui).

````python
class Persona:
    specie = "umano"

    def __init__(self, nome):
        self.nome = nome

    @classmethod
    def cambia_specie(cls, nuova_specie):
        cls.specie = nuova_specie

Persona.cambia_specie("extraterrestre")

# ORA: Tutte le persone ora hanno specie = "extraterrestre".
````

````python
class Persona:
    popolazione = 0             # Attributo di classe

    def __init__(self, nome):
        self.nome = nome        # Attributo di istanza
        Persona.popolazione += 1

    @classmethod
    def quante_persone(cls):
        return cls.popolazione

# Con il metodo di classe è possibile accedere e leggere l'attributo di classe
print(Persona.quante_persone())  # Output: 0
p1 = Persona("Alice")
print(Persona.quante_persone())  # Output: 1
````

#### 4.2. Metodi statici

Non hanno bisogno né dell’istanza (`self`) né della classe (`cls`).
- Decoratore: `@staticmethod`.
- Non possono leggere né modificare attributi di istanza o di classe, ma sono utili per funzioni collegate logicamente alla classe.

````python
class Matematica:
    @staticmethod
    def somma(a, b):
        return a + b

# Qui somma non usa né self né cls, ma è utile definirla all’interno della classe perché concettualmente appartiene a Matematica.
print(Matematica.somma(5, 3))  # Output: 8
````

#### 5.1. Eredità

Una classe può ereditarne un’altra: la classe figlia prende tutti gli attributi e metodi della classe madre; è utile per evitare duplicazioni.

````python
class Persona:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def saluta(self):
        print(f"Ciao, sono {self.nome}")

class Studente(Persona):
# Studente eredita da Persona.
# Questo significa che Studente prende automaticamente tutti i metodi e attributi di Persona (__init__, saluta, ecc.) senza doverli riscrivere.
    def __init__(self, nome, eta, scuola):
        super().__init__(nome, eta)  # richiama il costruttore della madre
        self.scuola = scuola

s2 = Studente("Marco", 18, "Liceo")
s2.saluta()  # Ciao, sono Marco
````

#### 5.2. Override


**L’override** serve quando vuoi modificare un metodo ereditato nella sottoclasse.

````python
class Studente(Persona):
    def __init__(self, nome, eta, scuola):
        super().__init__(nome, eta, scuola)
        self.scuola = scuola

    def saluta(self):
        # Qui saluta sostituisce (override) quello della classe madre Persona.
        print(f"Ciao, sono {self.nome} e studio a {self.scuola}")

s2 = Studente("Marco", 18, "Liceo")
s2.saluta()  # Ciao, sono Marco
````

#### 5.3. `super()`

    super().saluta()  # chiama saluta della madre

#### 5.4. Property

Le **property** permettono di trasformare un metodo in un attributo “calcolato” con getter e setter senza cambiare la sintassi dell’oggetto.

In Python, quando usi @property, crei un metodo che si comporta come un attributo in sola lettura:

````python
class Studente(Persona):
    def __init__(self, nome, eta, scuola):
        super().__init__(nome, eta)
        self.scuola = scuola

    @property
    def descrizione(self):
        return f"{self.nome}, {self.eta} anni, studia a {self.scuola}"

s2 = Studente("Marco", 18, "Liceo")
print(s2.descrizione)   # Qui s2.descrizione non si può assegnare, perché è solo un metodo di lettura
````

Ma se si vuole scrivere bisogna usare il decoratore `@<nome_property>.setter`,  e serve a definire un metodo che gestisce la scrittura di quell’attributo.

````python
class Studente(Persona):
    def __init__(self, nome, eta, scuola):
        super().__init__(nome, eta)
        self._scuola = scuola  # convenzione: attributo privato

    @property               # Getter
    def descrizione(self):
        return f"{self.nome}, {self.eta} anni, studia a {self._scuola}"

    @descrizione.setter     # Setter
    def descrizione(self, nuova_scuola):
        print("\n ... Sto aggiornando la scuola ...\n")
        self._scuola = nuova_scuola

s2 = Studente("Marco", 18, "Liceo")
print(s2.descrizione)  # Marco, 18 anni, studia a Liceo

s2.descrizione = "Istituto Tecnico"  # Viene passata il nuovo valore al metodo.setter, e di conseguenza aggiorna _scuola
print(s2.descrizione)  # Marco, 18 anni, studia a Istituto Tecnico
````

Il **deleter** è l’ultima parte del trio `@property` + `@setter` + `@deleter` e serve per cancellare un attributo in modo controllato, invece di farlo direttamente con del `obj.attributo`.

````python
class Studente(Persona):
    def __init__(self, nome, eta, scuola):
        super().__init__(nome, eta)
        self._scuola = scuola

    # GETTER
    @property
    def scuola(self):
        return self._scuola

    # SETTER
    @scuola.setter
    def scuola(self, nuova_scuola):
        if not nuova_scuola:
            raise ValueError("La scuola non può essere vuota!")
        self._scuola = nuova_scuola

    # DELETER
    @scuola.deleter
    def scuola(self):
        print(f"\n ... Cancellando la scuola di {self.nome} ...\n")
        self._scuola = None  # o del self._scuola se vuoi rimuoverlo completamente

# Esempio di uso
s2 = Studente("Marco", 18, "Liceo")
print(s2.scuola)  # Liceo

s2.scuola = "Istituto Tecnico"
print(s2.scuola)  # Istituto Tecnico

del s2.scuola      # chiama il deleter
print(s2.scuola)  # None
````

#### Metodi: Gestione di un Conto Bancario

**Obiettivi**

1) Ogni conto ha nome proprietario e saldo → attributi di istanza

2) La banca tiene traccia di numero totale conti → attributo di classe

3) Metodi:
    - `deposito(amount)` → metodo di istanza

    - `prelievo(amount)` → metodo di istanza

    - `mostra_saldo()` → metodo di istanza

    - `numero_conti()` → metodo di classe

    - `calcola_interesse(saldo, tasso)` → metodo statico

````python
class ContoBancario:
    n_conti_totali = 0      # Attributo di classe

    def __init__ (self, nome, saldo):
        self.nome = nome    # Attributo di istanza
        self.saldo = saldo  # Attributo di istanza
        ContoBancario.n_conti_totali += 1

    # 1 Metodo
    def deposito(self, amount):
        self.saldo += amount

    # 2 Metodo
    def prelievo(self, amount):
        self.saldo -= amount

    # 3 Metodo, con due opzioni
    def mostra_saldo(self):
        print(f"Il saldo disponibile è di {self.saldo} €")
    def __str__(self):
        return(f"Il saldo disponibile è di {self.saldo} €")

    # 4 Metodo di classe
    @classmethod
    def numero_conti_totali (cls):
        return cls.n_conti_totali

    # 5 Metodo statico
    @staticmethod
    def calcola_interessi (saldo, tasso):
        return saldo * tasso / 100
````

````python
c1 = ContoBancario("Luca", 100)
c1.deposito(50)
c1.prelievo(30)
print(c1)           # Usa il metodo speciale: __str__
c1.mostra_saldo()   # Usa il metodo d'istanza
````

````python
c1 = ContoBancario("Luca", 100)
c2 = ContoBancario("Anna", 200)

print("Il numero dei conti totali è di:",ContoBancario.numero_conti_totali())
print(f"Gli interessi per il conto c1 di {c1.nome} ammontano a: {c1.calcola_interessi(c1.saldo, 5)}%")
````

#### Sistema di utenti e prodotti

**Obiettivi**

1) Classe base: `Utente`

    - Attributi: `nome`, `email`

    - Metodo: `saluta()`

    - Property: `email` con getter e setter (es. controllo se contiene “@”)

2) Classe figlia: `Cliente` (eredita da `Utente`)

    - Attributo aggiuntivo: `carrello` (lista di prodotti)

    - Override metodo `saluta()` per dire “Ciao, sono cliente di XYZ”

3) Classe `Prodotto`

    - Attributi: `nome`, `prezzo`

    - Metodo: `sconta(tasso)` → restituisce prezzo scontato

4) Composizione:

    - `Cliente` contiene oggetti `Prodotto` nel carrello

    - Possiamo calcolare il totale del carrello

````python
class Utente:
    def __init__ (self, nome, email):
        self.nome = nome
        self._email = email
        # Essendo che posi si ha un metodo email, e self.email richiama di nuovo il getter → ciclo infinito
        # La soluzione è usare una variabile interna privata, di solito _email

    def saluta(self):
        print(f"Ciao, sono {self.nome}")

    @property
    def email(self):
        return self._email

    @email.setter
    def email(self, new_email):
        if "@" not in new_email:
            raise ValueError("L'email non è valida")
        self._email = new_email
````

````python
class Cliente(Utente):
    def __init__(self, nome, email):
        super().__init__(nome, email)  # inizializza nome e email
        self.carrello = []             # attributo di istanza per ogni cliente

    def saluta(self):
        print(f"Ciao, sono cliente di XYZ")

    def aggiungi_al_carrello(self, prodotto):
        self.carrello.append(prodotto)

    def totale_carrello(self):
        totale = 0
        for prodotto in self.carrello:
            totale += prodotto.prezzo
        return totale
````

````python
class Prodotto:
    def __init__(self, nome, prezzo):
        self.nome = nome
        self.prezzo = prezzo

    def sconta(self, tasso):
        return self.prezzo * (1 - tasso / 100)
````
